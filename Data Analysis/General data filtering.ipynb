```python
import os
import pandas as pd
```

```python
input_dir = "data"
output_dir = "Filtered datasets"
os.makedirs(output_dir, exist_ok=True)

exclude_companies = ['DB', 'Eu Sleeper', 'Eurobahn', 'NMBS', 'NS Int']
exclude_types = ['Stopbus ipv trein', 'Taxibus ipv trein', 'Bus']

months = {"01": "january", "02": "february", "03": "march", "04": "april", "05": "may", "06": "june", "07": "july", "08": "august", "09": "september", "10": "october", "11": "november", "12": "december"}

Saturdays = ['2024-01-06', '2024-01-13', '2024-01-20', '2024-01-27', '2024-02-03', '2024-02-10', '2024-02-17', '2024-02-24', '2024-03-02', '2024-03-09', '2024-03-16', '2024-03-23', '2024-03-30', '2024-04-06', '2024-04-13', '2024-04-20', '2024-04-27', '2024-05-04', '2024-05-11', '2024-05-18', '2024-05-25', '2024-06-01', '2024-06-08', '2024-06-15', '2024-06-22', '2024-06-29', '2024-07-06', '2024-07-13', '2024-07-20', '2024-07-27', '2024-08-03', '2024-08-10', '2024-08-17', '2024-08-24', '2024-08-31', '2024-09-07', '2024-09-14', '2024-09-21', '2024-09-28', '2024-10-05', '2024-10-12', '2024-10-19', '2024-10-26', '2024-11-02', '2024-11-09', '2024-11-16', '2024-11-23', '2024-11-30', '2024-12-07', '2024-12-14', '2024-12-21', '2024-12-28']

dfs = [] 

for num, name in months.items():
    input_path = os.path.join(input_dir, f"services-2024-{num}.csv.gz")
    output_path = os.path.join(output_dir, f"traindata_{name}_filtered.csv")

    df = pd.read_csv(input_path, compression='gzip')

    df = df[df['Service:Date'].isin(Saturdays)]

    df = df[~df['Service:Company'].astype(str).str.strip().isin(exclude_companies)]
    df = df[~df['Service:Type'].astype(str).str.strip().isin(exclude_types)]

    df.to_csv(output_path, index=False)
    dfs.append(df)


combined_df= pd.concat(dfs, ignore_index=True)
combined_output_path = os.path.join(output_dir, "traindata_filtered_2024.csv")
combined_df.to_csv(combined_output_path, index=False)   
#To be able to use the combined_df in other script of disruption analysis, saved as pickle file 
combined_df.to_pickle("combined_df_zaterdagen.pkl")
```
