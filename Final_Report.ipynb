### Resilience Dutch train network and large events  · 6 November 2026

## 1. Introduction
Large-scale events such as King’s Day attract thousands of visitors to major Dutch cities, placing additional pressure on the railway network. Passenger volumes increase, travel patterns become concentrated and more occasional travellers use the system. To accommodate this demand, railway operators may adjust timetables and service frequencies. Understanding how the network responds to these challenges is essential for assessing its operational resilience. This project therefore investigates the impact of King’s Day 2024 on Dutch railway performance.

## 2. Research objective
The objective of this research is to assess the operational resilience of the Dutch railway network during King’s Day by analysing changes in train punctuality, delays, cancellations, passenger volumes and service frequency. By comparing railway performance on 27 April 2024 **with a comparable Saturday one week earlier)**, the study aims to identify the operational impact of the event and determine which stations and cities are most affected.

## 3. Research question
How resilient is the Dutch railway network's operational performance, measured by train punctuality, cancellations, and delays, during major national events and planned capacity shifts? 

**Subquestion 1:**
-	To what extent does the number of train delays during King’s Day differ from the average number of delays observed at Dutch railway stations?

**Subquestion 2:**
-	How do train capacity and service frequency change on King’s Day compared to an average day

**Subquestion 3:**
-	How does the timetable/schedule change in comparison to an average day?

**Subquestion 4:**
-	To what extent does passenger volume on King’s Day differ from that on an average day?

**Subquestion 5:**
-	How do train delays during King’s Day differ between cities in the Netherlands?
