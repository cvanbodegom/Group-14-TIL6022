## 1. Introduction
Large-scale events such as King’s Day attract thousands of visitors to major Dutch cities. Around 270.000 people take the train to Amsterdam on King’s Day, placing additional pressure on the railway network (NOS Nieuws, 2014). Passenger volumes increase, travel patterns become concentrated and more occasional travellers use the system. To accommodate this demand, railway operators may adjust timetables and service frequencies. Understanding how the network responds to these challenges is essential for assessing its operational resilience. The performance of the Dutch railway system is based on the percentage of passengers for whom the train trip was successful in terms of travel time (CLO, 2024).This project therefore investigates the impact of King’s Day 2024 on Dutch railway performance.

## 2. Research objective
The objective of this project is to examine how resilient the Dutch railway network is during King’s Day, one of the largest national events in the Netherlands. As this event generates distinct travel patterns compared with a typical day, it provides an opportunity to examine how the railway network performs under increased passenger demand. This is investigated by comparing train delays, measured in total minutes of delay, on King’s Day 2024 with those recorded on a typical Saturday. This comparison aims to determine whether the increased travel demand associated with King’s Day affects railway performance, particularly in terms of arrival delays.

## 3. Research question

### 3.1 Main research question

How resilient is the Dutch railway network's operational performance, measured by train delays and cancellations during King’s Day?

In this study, resilience is understood as the ability of the railway network to absorb a sudden increase in passenger demand while continuing to run its services (few cancellations and disruptions) and keeping train delays close to normal levels.

### 3.2 Sub questions
1.	To what extent does the total arrival delay of trains on King’s Day differ from that on an average Saturday in 2024, and can this difference be explained by extreme delays, cancellations or disruptions?
2.	How does the service frequency and timetable change on King’s Day compared to an average day?
3.	To what extent does passenger volume on King’s Day differ from that on an average day?
4.	How do train delays during King’s Day differ between cities in the Netherlands?





## 4. Data Set

For this research, three different datasets are used.
1.	A large open dataset which has archived all train rides in the Netherlands. This open data set contains all passenger train services in the Netherlands for a month from the year 2019. The data set can be obtained from Rijden de Treinen (Open data over treinritten, n.d.)
2.	Another dataset from Rijden de Treinen, which contains all the train disruptions since 2011 (Open data over treinritten, n.d.). 
3.	Data set about the number of check-ins using OVChipkaart. This dataset is from Translink (Translink | Open data, n.d.)

### 4.1 Delay time data 
The data on the delays of trains in the Netherlands can be found in the dataset about all train rides in the Netherlands. This dataset is used to answer the sub questions, as the performance of the Dutch railway network is evaluated based on delays. Each row in this dataset represents a stop at a station. In this research, the data from 2024 is used. The dataset includes information about the following:

-	The type of train and carrier performing the trip (Service: Type & Service:Company) 
-	If the train is completely or partly cancelled (Service: Completely cancelled & Service: Partly Cancelled)
-	The maximum delay for the ride (Service:Maximum delay)
-	The name of the station (Stop:Station name)
-	The delay on arrival in minutes (Stop:Arrival delay)

### 4.1.1 Data cleaning 
The twelve different datasets(services-2024-xx.csv.gz) for each month from Rijden de Treinen Open Data are used for the analysis. To ensure a representative comparison of train delays on King’s Day (27 april 2024), only Saturday observations are included, as King’s Day fell on a Saturday and previously identified differences in travel behaviour show a difference in travel patterns between weekdays and weekends. Therefore, all the 52 Saturdays in 2024 are selected. 
Second, the datasets also contain international trains or busses that are a replacement for the trains, because this report looks at the resilience of the Dutch train network the following types that are in the column of “Service:Type” and “Service: Company” are excluded: 
1. International railway operators: ‘DB’, ‘Eurostar’, ‘Eurobahn’, ‘NMBS’, and ‘NS International’ were excluded to focus the analysis on the relevant domestic railway services.
2. Replacement bus services: ‘Stopbus ipv trein’ was excluded because the research focuses on train services rather than replacement transport.
3. Taxi bus services: ‘Taxibus ipv trein’ was excluded for the same reason
4. Regular bus services: ‘Bus’ was excluded because the analysis focuses on railway traffic rather than bus transport.
Each monthly dataset is filtered to retain only records from Saturdays and excluded certain transport types. The resulting datasets are combined into a single dataset and exported as traindata_filtered_2024, which serves as the input for the different analysis. 

### 4.2 Train disruptions data 
The data on the disruptions contains records of the different disruptions trains experienced between 2011 and 2025. In this research only the data from the year 2024 is used. The dataset includes the following information:

-	The affected routes (ns_lines)
-	The cause of the disruption (cause_nl)
-	The group in which the disorder is classified (cause_group)
-	The duration of the disruption in minutes (duration_minutes)
-	The time and date when the disruption occurred (start_time)

In this research the disruption dataset of 2024, which consists of  5965 observations, is used. The raw dataset is left unmodified, as in this research the python code is used to select the relevant observations for the research questions. 

### 4.3 Number of passengers data 
The number of passengers datasets from Translink contains information about the check-ins on all public transport (train, bus, tram and metro) in 2024. The dataset contains three columns:
-	Date 
-	Hour 
-	Number of Check-ins


As the dataset is relatively small, all observations are retained for the analysis, and no data points are excluded.


## 5. Data Analysis

### 5.1 To what extent does the total arrival delay of trains on King’s Day differ from that on an average Saturday in 2024, and can this difference be explained by extreme delays, cancellations or disruptions?
To investigate whether King's Day had an impact on railway performance, the total train delays recorded on 27 April 2024 are compared with those observed on other Saturdays throughout 2024. This comparison provides insight into whether the level of train delays on King's Day was unusually high compared with the average Saturday. In addition to comparing total delays, several potential explanations for the observed differences are examined. These include the influence of exceptionally long train delays, train cancellations, and the duration of railway disruptions. This analysis helps to determine whether the higher total delay on King's Day can be linked to any of these factors.


![Figure 1](../Figures/figure1.png)

*Figure 1: Total train delay per Saturday in 2024*

Figure 1 shows the total train delay, measured in minutes, for each Saturday in 2024 from January to December. Each bar represents the cumulative delay across all included train services on a particular Saturday.
The main observation is that total train delays fluctuate considerably throughout the year, ranging from approximately 23,000 to 55,000 minutes. The orange bar represents 27 April 2024 (King’s Day), when the total delay amounted to approximately 49,500 minutes. This is considerably higher than the average total delay of 37,162 minutes, calculated across all Saturdays remaining after applying the data filters. The total delay on King’s Day was 32.75% higher than the average for a typical Saturday. Furthermore, only three other Saturdays recorded higher total delays than King’s Day, indicating that the level of disruption on this date was relatively high compared with other Saturdays in 2024.



![Figure 2](../Figures/figure2.png)

*Figure 2: Total train delay per Saturday in 2024, with delays above 60 minutes excluded (top) and all delays included (bottom)*

Figure 2 compares the total train delays per Saturday before and after excluding individual train delays exceeding 60 minutes. The upper chart shows the results after excluding these extreme delays, while the lower chart displays the original data. The comparison shows that removing delays exceeding 60 minutes reduces the total delay for some Saturdays, but the overall pattern remains largely unchanged. In particular, the total delay on King's Day remains relatively high compared with most other Saturdays. This suggests that the relatively high total delay on King's Day cannot be explained solely by a small number of exceptionally long train delays. Therefore, the original dataset is retained for the further analysis.
To investigate the possible causes of the higher total train delays on King's Day, the number of cancelled trains on each Saturday was examined. This analysis aimed to determine whether train cancellations on King's Day could have significantly contributed to the higher total delay. The results are presented in Figure 3.


![Figure 3](../Figures/figure3.png)

*Figure 3: Number of unique cancelled trains per Saturday*

As shown in Figure 3, King's Day (highlighted by the orange date label) had a relatively low number of cancelled trains. The figure distinguishes between fully cancelled trains and partially cancelled trains, where a train does not complete its full route and stops at an earlier station. This suggests that the higher total train delays on King's Day were not primarily caused by train cancellations.
To investigate whether the high total train delay on King's Day was related to disruptions taking a long time to resolve, the total train delay was compared with the total disruption duration for each Saturday. The results are presented in Figure 4.


![Figure 4](../Figures/figure4.png)

*Figure 4: Total delay vs. total disruption time per Saturday in 2024*

As shown in Figure 4, the total train delay on King's Day was relatively high, as discussed earlier. However, the total disruption duration on this date was relatively low compared with that of several other Saturdays. This suggests that the high total train delay on King's Day was not primarily associated with disruptions lasting an unusually long time. Other factors may therefore have contributed to the higher total delay.
Overall, the analysis shows that the total train delay on King's Day was considerably higher than the average across the other Saturdays in the filtered dataset. The additional analyses suggest that this difference cannot be explained solely by exceptionally long individual train delays, train cancellations, or unusually long disruption durations. 


### 5.2 How does the service frequency and timetable change on King’s Day compared to an average day?
This section will describe how the service frequency and timetable changes on King’s Day compared to an average day. The average day is once again the average over all Saturdays in the year 2024. This sub question will be answered by utilizing the Open Data of RijdendeTreinen.nl. 


![Figure 5](../Figures/figure5.png)

*Figure 5: Train traffic comparision by city cluster: Average Saturday vs. King's Day by city cluster* 

Figure 5 compares total train volumes across major city clusters on King’s Day versus an average Saturday baseline (King’s Day is excluded in this baseline). The selection covers major city hubs in the Randstad alongside key regional centres, with Emmen being included due to the hosting of the royal visit on April 27th. 
The comparison reveals a distinct distribution of trains across the network. First, Amsterdam shows dominance with massive number of trains added due to the biggest and most popular events happening here. Under the specialised Oranjedienstregeling, NS increases train services to Amsterdam to accommodate the expected influx of passengers during King’s Day (Nederlandse Spoorwegen, 2024). The Oranjedienstregeling uses longer trains, but fewer trains in total (Connect, n.d.). Across all stations in the dataset, the number of trains on King’s Day was 9.8% lower than on an average Saturday. Another reasoning is the fact that King’s Day is an official holiday meaning that people do not need to go to work (Connect, n.d.). 
Conversely, major connecting hubs such as Utrecht, Rotterdam, and Den Haag record notable drops in train volume. This reduction shows a deliberate trade-off rather than loss of network functionality. The ‘Sprinter’ trains (regional loops) are thinned out. This capacity is relocated to support high-capacity intercity corridors and longer train formations. The train length formation is maximized. In addition, some stations are closed to prevent congestions. In Amsterdam, these stations are; Amsterdam RAI and Amsterdam Science Park. Eindhoven does not show a lot of change, this is expected since Eindhoven is a self-contained regional hub. Emmen however shows an increase in traffic which is driven by the visit of the royal family. 


![Figure 6](../Figures/figure6.png)

*Figure 6:Hourly train frequency profile: King's Day vs. average Saturdays*

Figure 6 illustrates the hourly train departure profile across all target stations from 00:00 to 23:00 on King’s Day compared to an average Saturday baseline. Roughly around 03:00-05:00 train departures rise slightly above the average Saturday. This is likely driven by repositioning runs to deploy rolling stock to their positions in advance of the service. Contrary to what might be expected during a major event, the train volume between 06:00-09:00 is below the Saturday average. Operators suppress or delay standard service runs during these hours. In addition, people do not need to work during this holiday, and people might sleep in more due to King’s Night before. The rest of the day till 19:00 develops in a standard manner. After this time, frequencies drop off much more sharply than on a typical Saturday. This is expected to reflect the rapid wind-down of the event timetables as crowds disperse and the network transitions back toward standard night schedules. 

![Figure 7](../Figures/figure7.png)

*Figure 7: Train traffic percentage growth on King’s Day*

Figure 7 illustrates the station-level percentage growth in train traffic relative to a standard Saturday. The station percentages reflect the before mentioned trade-offs and relations with the targeted operational interventions. Primary nodes in the event and transfer stations see an increase in traffic, such as Amsterdam Holendrecht (+71.9%), Amsterdam Centraal (+36.4%) and Amsterdam Muiderpoort (+35.0%). Passengers are funnelled directly into the high-intensity zones within Amsterdam. Secondary and localised stops show dramatic reductions. Most notably Amsterdam Science Park (-99.1%) and Amsterdam RAI (-54.7%) experiences steep declines due to closures. However, these do not drop to 100% due to operational runs. This traffic is consistent with equipment repositioning as well or underlying timetable blocks that remain logged. 
Overall, King’s Day results in a redistribution of train services across the Dutch railway network. More trains are directed towards key locations such as Amsterdam and Emmen, while other major hubs experience a reduction in train frequency. In addition, longer trains and station closures are used to manage the expected passenger demand and prevent overcrowding. This shows that the King’s Day timetable focuses more on passenger capacity and targeted services than on increasing train frequency across the entire network. King’s Day shifts from a frequency-based to a capacity-based timetable.


### 5.3 To what extent does passenger volume on King’s Day differ from that on an average day?
Passenger volume is measured using Translink check-in data, which records the number of check-ins per hour across Dutch public transport. Because the data covers all modes (train, bus, tram and metro), it reflects overall public transport demand rather than rail demand alone. It is nevertheless a suitable indicator of the extra travel demand generated by King’s Day.


![Figure 8](../Figures/figure8.png)

*Figure 8: Total check-ins per Saturday in 2024 (x1,000)*

On King’s Day 2024, about 3.36 million check-ins were recorded, 37.5% more than the average of the other Saturdays in 2024 (2.45 million). This made it the busiest Saturday of the year, as you can see by the orange bar in the chart in figure 8. The hourly profile in figure 9 shows that the largest difference occurs around midday, with check-ins peaking at about 325,000 per hour at 13:00, compared with about 210,000 on other April Saturdays. In the evening, demand also remains clearly higher than on a regular Saturday, when check-ins gradually decline. 

![Figure 9](../Figures/figure9.png)

*Figure 9: Hourly check-ins on King's Day compared with the average of other April Saturdays[SB2.1] (x1,000)*

A likely explanation is that many passengers travel to major cities to join the celebrations (NOS, 2024). They depart later in the morning and return after the festivities, concentrating passenger flows around midday and into the evening.
King’s Day thus placed an exceptional demand on the network, concentrated in a few hours of the day. Since Section 5.2 showed that fewer trains ran than on an average Saturday, more passengers were carried by fewer, longer trains. The next section examines whether this is reflected in train delays across Dutch cities.

### 5.4 How do train delays during King’s Day differ between cities in the Netherlands? 
To examine whether the impact of King’s Day was consistent across the country, the number of minutes in train delays for each Saturday were compared between six Dutch cities and their respective averages. Emmen recorded the largest average increase at 64.7% most likely due to the royal celebration being held there. This causes an unnatural surge compared to normal Saturdays in Emmen even with the NS 'oranjedienstregeling'. Amsterdam also sees a significant increase at 32.3%. This is due to the greater volumes of people traveling in and out of Amsterdam where the biggest collection of celebrations and events is.

These differences can be considered alongside the changes in train services described in Section 5.2. In Amsterdam and Emmen, where train volumes increased, total delay also increased. In contrast, Utrecht, Rotterdam and Den Haag experienced reductions in service frequency and recorded a decrease in total delay. This suggests that the change in total delay is strongly influenced by the number of trains running. Eindhoven is an exception: total delay increased substantially while train volume remained almost unchanged. This indicates that train volume alone does not explain all differences in delays. Concentrated passenger flows, adjusted timetables and local operating conditions may also have played a role. Section 5.5 examines this further by correcting the delays for train volume.


![Figure 10](../Figures/figure10.png)

*Figure 10: Relative change in total delay minutes*

While looking at the hourly distribution of train delays per city we noticed a similar patters as the number of check ins per hour of the day as mentioned in section 5.3. This is best seen in Amsterdam, which of course has the highest share of passengers and thus shares the most direct resemblance for the uniquely delayed trains most likely caused by the greater volume of passengers. But when closely inspecting the other cities the volume of passengers according to the check ins in 5.3 do approach the similar peak moments.

![Figure 11](../Figures/figure11.png)

*Figure 11: Distribution of uniquely delayed trains per big city on King’s Day*

It is important to distinguish between the absolute number of train delays and the relative change compared with an average Saturday. The reduction in train frequency observed in Utrecht, Rotterdam, and The Hague may have several explanations. Fewer train services mean fewer individual trains that can experience delays, while longer trains provide greater passenger capacity per departure. This may reduce overcrowding and boarding times, potentially helping to limit delays. Another explanation is that more people celebrate King’s Day locally rather than travelling to other cities, resulting in different passenger flows across the network.
However, these explanations cannot be confirmed using the current data alone. Further qualitative research, for example through a national survey, could investigate why people choose to travel or stay in their own city on King’s Day. This would provide additional insight into how travel behavior influences railway demand and operational performance during large-scale events.



### 5.5 Combining train volume and delays
The previous sections examined passenger demand, service frequency, and delays separately. However, on a day with more running trains, it is more likely that there will be more delays. Section 5.2 shows that the number of trains on King's Day was not equal to that of an average Saturday. Comparing total delays alone may therefore misrepresent how much the performance of individual trains has changed. 


![Figure 12](../Figures/figure12.png)

*Figure 12: Change in train volume, total delay and delay per train on King’s Day 2024 compared with a average Saturday per city cluster.*

Figure 12 compares the change in train volume, total delay and delay per train for each city cluster. Correcting for train volume changes the picture considerably. In Amsterdam and Emmen, part of the increase in total delay is explained by the extra trains that were scheduled. Delay per train still increased, by 13.0% in Amsterdam and 23.7% in Emmen. In Eindhoven, train volume was almost unchanged, so the increase in delay per train (17.7%) closely matches the increase in total delay.
The most notable results are found in Den Haag and Utrecht. Both cities recorded a decrease in total delay, but train volume decreased even more strongly (−24.8% and −16.7%). As a result, delay per train actually increased by 14.2% in Den Haag and 9.3% in Utrecht. Rotterdam is the only city where delay per train decreased (−11.9%).
Overall, delay per train increased in five of the six cities. The effect of King’s Day on delays was therefore not limited to Amsterdam but was visible across the network.



## 6 Conclusion
This research looked at quantitative data on the relation between a national holiday in the Netherlands and the train delays and cancellations in the rail network. The chosen day was King’s Day 2024. The data on this day was compared with all other Saturdays in the same year. This led to the following research question:
How resilient is the Dutch railway network’s operational performance, measured by train delays and cancellations, during King’s Day?
Looking at the results of the sub-questions, the Dutch rail network showed mixed resilience on King’s Day compared with other Saturdays in the year. Below, the individual findings are discussed.
Firstly, with 37.5% more check-ins than an average Saturday, King’s Day 2024 was the busiest Saturday of the year (Section 5.3). It also scored the fourth highest in total delay minutes of all Saturdays (Section 5.1). NS is aware of the increase in passenger volume, which is why it prioritises the capacity of the trains by using longer trains (Nederlandse Spoorwegen, 2024). The frequency of trains does not change much; it even decreases slightly, by 9.8%, compared to an average Saturday (Section 5.2).
King’s Day also performs comparatively well in the number of cancellations. It is one of the Saturdays with the lowest number of cancelled trains. This is possibly due to preparations for malfunctions and other causes of cancellations. The network’s ability to absorb such a surge in passenger volume without more cancellations highlights its operational resilience in terms of continuity.
However, the extra delay was not limited to the busiest parts of the Netherlands. When corrected for the number of trains, delay per train increased in five of the six cities analysed, including Den Haag (+14.2%) and Utrecht (+9.3%), where fewer trains ran (Section 5.5).
In conclusion, the Dutch railway network’s operational performance was resilient in terms of cancellations, as services kept running with few cancelled trains. In terms of delays, however, performance did not stay close to normal levels: total delay was 32.75% higher than on an average Saturday, and delay per train increased across most of the network.



### 6.1 Limitations 
While the findings of this research provide insight into disruption of data for the rail network, several limitations must be addressed. Starting with the fact that this research covers the data of only one event day and a chosen set of train stations, possibly excluding some interesting variances. The usual train delays are already widely spread, between 23,000 and 55,000 minutes of delay this year. For a special day such as King’s Day, the data can also vary greatly over multiple years due to unforeseen circumstances. Comparing multiple years would make the results more reliable and less sensitive to one-off circumstances.
Besides, the check-in data that was used is the national number of check-ins, because data for each station and for each public transport mode separately was not available for this research. As a result, the number of passengers could not be linked directly to a city, but only to the increased volume of the network as a whole.
In addition, the dataset contains no information on train length or composition. The use of longer trains during the Oranjedienstregeling could therefore not be verified with the data used in this study. Replacement bus services were also excluded from the analysis, while extra buses were deployed on King’s Day (Connect, n.d.). Part of the reduction in train services may therefore have been compensated for by buses, which is not reflected in this study.
For the delay data, it can give a distorted view that delays are counted at every stop. If a train is delayed by 10 minutes at the first stop, it might still be 8 minutes delayed five stations later, because of the maximum speed it is allowed to travel on the route. This causes a streak of delays that, when summed up, becomes considerably high over a prolonged period of time. These delays therefore reflect not only delays due to traffic or other circumstances, but also the route length and the ability to make up time. This can also be a reason why Amsterdam has a considerably higher total amount of delay time, as it has most stations where trains can log an extra delay when arriving at the station. On the other hand, the small number of trains in Emmen makes the delay percentage look disproportionate, as a couple of disruptions can cause a bigger percentage increase compared with normal circumstances.
Another issue is external factors that are not accounted for, such as weather disruptions. These can be found online individually but are excluded from this study. A follow-up study could account for these to create a more accurate dataset.
Lastly, the results suggest that the high number of passengers travelling by train is the main cause of train delays, but this cannot be confirmed, as the causes of delays are not included in the data. In addition, official punctuality is measured per passenger (CLO, 2024), while this study measures delay per train. Passenger-related delays can have a number of different reasons. For example, train carriages can be overloaded beyond their usual capacity, or issues between passengers or train staff can cause a train to be halted. These effects cannot be confirmed with the available data and must be researched in a different way.
