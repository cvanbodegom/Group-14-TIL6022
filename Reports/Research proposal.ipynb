# Data Used

A large open dataset which has archived all train rides in the Netherlands. This open data set contains all passenger train services in the Netherlands for a month from the year 2019. The data set can be obtained from Rijden de Treinen (https://www.rijdendetreinen.nl/open-data/treinarchief) with 12 sets out of 2024 where xx would be the month; services-2024-xx.csv.gz 

We will use the following parts within this dataset;
-	The type of train and carrier performing the trip (Service: Type & Service:Company) 
-	If the train is completely or partly cancelled (Service: Completely cancelled & Service: Partly Cancelled)
-	The maximum delay for the ride (Service:Maximum delay)
-	The name of the station (Stop:Station name)
-	The delay on arrival in minutes (Stop:Arrival delay)

In addition we will not use all parts; 
Second, the datasets also contain international trains or busses that are a replacement for the trains, because this report looks at the resilience of the Dutch train network the following types that are in the column of “Service:Type” and “Service: Company” are excluded: 
-	International railway operators: ‘DB’, ‘Eurostar’, ‘Eurobahn’, ‘NMBS’, and ‘NS International’ were excluded to focus the analysis on the relevant domestic railway services.
-	Replacement bus services: ‘Stopbus ipv trein’ was excluded because the research focuses on train services rather than replacement transport.
-	Taxi bus services: ‘Taxibus ipv trein’ was excluded for the same reason
-	Regular bus services: ‘Bus’ was excluded because the analysis focuses on railway traffic rather than bus transport.

Another dataset from Rijden de Treinen, which contains all the train disruptions since 2011 (https://www.rijdendetreinen.nl/open-data/treinstoringen). In this research the data out of 2024 will be used only. With the specific name of the data set being; disruptions-2024.csv
We will use the following parts within this dataset;
-	The affected routes (ns_lines)
-   The cause of the disruption (cause_nl)
-	The group in which the disorder is classified (cause_group)
-	The duration of the disruption in minutes (duration_minutes)
-   The time and date when the disruption occurred (start_time)


Data set about the number of check-ins using OVChipkaart. This dataset is from Translink (https://translink.nl/open-data/) with the specific set being; Instapers_per_uur_export_V3_2024.csv
We will use the following parts within this dataset;
-	Date 
-	Hour 
-	Number of Check-ins
